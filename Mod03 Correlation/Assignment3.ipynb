2. Create a dataframe from the data contained in *income_data.csv* using `pandas`

```python
import pandas as pd
from scipy import stats

df = pd.read_csv('./income_data.csv')
df.head()
```

3. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *income* and *education*

```python
# Income and Education
print('Income and Education')
r, p = stats.pearsonr(df['income'], df['education'])
print(f'r (Pearson correlation coefficient): {r:.3f}')
print(f'p-value:                             {p:.3f}')

print('')
```

4. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *income* and *prestige*

```python
# Income and Prestige
print('Income and Prestige')
r, p = stats.pearsonr(df['income'], df['prestige'])
print(f'r (Pearson correlation coefficient): {r:.3f}')
print(f'p-value:                             {p:.3f}')

print('')
```

5. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *education* and *prestige*

```python
# Education and Prestige
print('Education and Prestige')
r, p = stats.pearsonr(df['education'], df['prestige'])
print(f'r (Pearson correlation coefficient): {r:.3f}')
print(f'p-value:                             {p:.3f}')

print('')
```
