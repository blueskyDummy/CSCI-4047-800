```python
# Code here
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('./Income Data.csv')

df.head
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
# Code here

def fns(column):
    min = np.min(df[column]) 
    q1 = np.percentile(df[column], 25)
    med = np.median(df[column])
    mean = np.mean(df[column])
    q3 = np.percentile(df[column], 75)
    max = np.max(df[column])

    print(f'------------ FNS for {column.capitalize()} -------------')
    print(f"Min: {min}")
    print(f"Q1: {q1}")
    print(f"Median: {med}")
    print(f"Mean: {mean}")
    print(f"Q3: {q3}")
    print(f"Max: {max}")

fns('income')
fns('education')
fns('prestige')
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
# Code here
plt.hist(df['prestige'], bins=12, color='orange', alpha=0.5)
plt.show()

plt.hist(df['education'], bins=12, color='purple', alpha=0.5)
plt.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
# Code here
plt.scatter(x=df['income'], y = df['education'], alpha=0.3)
plt.title("Income compared to education")
plt.xlabel('Income')
plt.ylabel("Education")
plt.show()

plt.scatter(x=df['education'], y = df['prestige'], alpha=0.3)
plt.title("Education compared to prestige")
plt.xlabel('Education')
plt.ylabel("Prestige")
plt.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
# Code here
standard_dev = df['income'].std()
print()
print(f"Income Standard Deviation:  {standard_dev:.2f}")

variance = standard_dev ** 2 # x ** y is the same as 'x to the power of y'
print(f"Income Variance:            {variance:.2f}")
```
