2. Create a dataframe of the data contained in **Income Data.csv** using `pandas`. You may use the code from Assignment 1 (assuming it works :P).

```python
#Creating dataframe
df = pd.read_csv("Income Data.csv")
df.head()
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
# Alternative method
# print(df.describe())
#------------------INCOME------------------
min = np.min(df.income)
q1 = np.percentile(df.income, 25)
med = np.median(df.income)
mean = round(np.mean(df.income), 2)
q3 = np.percentile(df.income, 75)
max = np.max(df.income)

print('------------ INCOME DATA -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Mean: {mean}")
print(f"Q3: {q3}")
print(f"Max: {max}")

#-----------------EDUCATION--------------
min = np.min(df.education)
q1 = np.percentile(df.education, 25)
med = np.median(df.education)
mean = round(np.mean(df.education), 2)
q3 = np.percentile(df.education, 75)
max = np.max(df.education)

print('------------ EDUCATION DATA -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Mean: {mean}")
print(f"Q3: {q3}")
print(f"Max: {max}")

#-----------------PRESTIGE--------------
min = np.min(df.prestige)
q1 = np.percentile(df.prestige, 25)
med = np.median(df.prestige)
mean = round(np.mean(df.prestige), 2)
q3 = np.percentile(df.prestige, 75)
max = np.max(df.prestige)

print('------------ PRESTIGE DATA -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Mean: {mean}")
print(f"Q3: {q3}")
print(f"Max: {max}")
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
#--------PRESTIGE HISTOGRAM--------
print('------------ PRESTIGE HISTOGRAM -------------')
plt.hist(df['prestige'], bins=7, color='red', alpha=0.5)
plt.show()
#--------EDUCATION HISTOGRAM--------
print('------------ EDUCATION HISTOGRAM -------------')
plt.hist(df['education'], bins=7, color='red', alpha=0.5)
plt.show()

```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
#--------INCOME & EDUCATION SCATTER PLOT--------
plt.scatter(df['income'], df['education'], alpha=.3)
plt.title("Income to Education")
plt.xlabel('income')
plt.ylabel("education")
plt.show()
#--------EDUCATION & PRESTIGE SCATTER PLOT--------
plt.scatter(df['education'], df['prestige'], alpha=0.3)
plt.title("Education to Prestige")
plt.xlabel('income')
plt.ylabel("prestige")
plt.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
#Variance
in_var= round((np.var(df['income'])), 2)
print(f"Income variance: {in_var}")
#Standard Deviation 
in_sd = round((np.std(df['income'])),2)
print(f"Income Standard Deviation: {in_sd}")
```
